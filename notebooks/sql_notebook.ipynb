```python
%sql
CREATE VIEW IF NOT EXISTS oil_analytics.oil_market_daily_final AS
SELECT 
om.date, 
brent_crude_usd, 
brent_return, 
brent_7d_avg,
brent_30d_avg,
wti_usd,
wti_return,
wti_7d_avg,
wti_30d_avg,
natural_gas_usd,
natgas_return,
natgas_7d_avg, 
natgas_30d_avg, 
brent_wti_spread,
sp500_close,
sp500_return,
dxy_close,
dxy_return, 
brent_vs_sp500_corr_30d, 
wti_vs_sp500_corr_30d,
brent_vs_dxy_corr_30d,
wti_vs_dxy_corr_30d,
uk_inflation_rate_1m,
us_inflation_rate_1m,
us_inflation_12m_avg,
uk_inflation_12m_avg,
us_high_inflation_regime,
uk_high_inflation_regime,
uk_unemployment_rate_1m,
us_unemployment_rate_3m,
uk_unemployment_12m_avg,
us_unemployment_12m_avg,
uk_weak_labour_market_flag,
us_weak_labour_market_flag,
uk_interest_rate_3m,
us_interest_rate_1m,
uk_rate_hike_flag,
us_rate_hike_flag,
uk_rate_cut_flag,
us_rate_cut_flag,
uk_rate_hold_flag,
us_rate_hold_flag,
uk_gdp_rate_yoy,
us_gdp_rate_yoy,
uk_weak_growth_regime,
us_weak_growth_regime
FROM oil_analytics.gold_oil_market_features om
LEFT JOIN oil_analytics.macro_inflation_daily inf 
ON om.date = inf.date
LEFT JOIN oil_analytics.macro_unemployment_daily unem
ON om.date = unem.date
LEFT JOIN oil_analytics.macro_interest_daily inter
ON om.date = inter.date
LEFT JOIN oil_analytics.macro_gdp_daily gdp
ON year(om.date) = gdp.year
ORDER BY om.date DESC
```
