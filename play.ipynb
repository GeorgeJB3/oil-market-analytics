```python
def fetch_dollar_index_data():

    dollar_index = yf.Ticker("SPY")
    dollar_index_df = dollar_index.history(period="5y", interval="1d", auto_adjust=True)
    dollar_index_df = dollar_index_df.reset_index()
    return dollar_index_df
```
